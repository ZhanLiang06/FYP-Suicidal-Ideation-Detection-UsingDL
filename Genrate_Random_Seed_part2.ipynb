```python
import random
from datetime import datetime

cur_datetime = datetime.now()
timestamp_float = cur_datetime.timestamp()
timestamp_int = int(timestamp_float)
print(f"Current datetime: {cur_datetime}")
print(f"Timestamp (float): {timestamp_float}")
print(f"Timestamp (int): {timestamp_int}")

random.seed(timestamp_int)

# Variable to control the number of seeds to generate
num_seeds_to_generate = 20

generated_seeds = []
for i in range(num_seeds_to_generate):
    # Generate a random seed between 0 and 2^32 - 1
    seed = random.randint(0, 2**32 - 1)
    generated_seeds.append(seed)

print(f"Generated {num_seeds_to_generate} random seeds: {generated_seeds}")

generated_seeds.sort()
print(f"Sorted seeds: {generated_seeds}")
```
